```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("sports_data_base_dav_1_project(1).csv")

print("Rows and columns:", df.shape)

df.head()
```

```python
df.info()

```

```python
# Check exact duplicate rows
print("Exact duplicate rows:", df.duplicated().sum())
```

```python
# Check duplicate IDs
dupe_mask = df.duplicated(subset=["ID"], keep=False)
print("Rows involved in duplicate IDs:", dupe_mask.sum())

if dupe_mask.sum() > 0:
    display(
        df[dupe_mask]
        .sort_values("ID")
    )
else:
    print("No duplicate IDs found.")

```

```python
# Remove duplicate IDs
before = len(df)

df = df.drop_duplicates(
    subset=["ID"],
    keep="first"
).reset_index(drop=True)

after = len(df)

print(f"Records removed: {before - after}")
print("Rows now:", after)
```

```python
# Display missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python
missing_count = df.isnull().sum()

missing_percent = (
    df.isnull().sum() / len(df)
) * 100

missing_report = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percent
})

print(missing_report)
```

```python
numeric_cols = [
    "Age",
    "Overall",
    "Potential",
    "Finishing"
]

for col in numeric_cols:

    print(col, "skewness:", df[col].skew())

    plt.figure(figsize=(6, 4))

    plt.hist(
        df[col].dropna(),
        bins=20,
        edgecolor="black"
    )

    plt.title("Distribution of " + col)
    plt.xlabel(col)
    plt.ylabel("Frequency")

    plt.show()
```

```python
# Position distribution
df["Position"].value_counts().plot(
    kind="bar",
    figsize=(10, 5)
)

plt.xlabel("Position")
plt.ylabel("Number of Players")
plt.title("Players by Position")
plt.xticks(rotation=45)

plt.show()
```

```python
# Fill numerical missing values using median
for col in numeric_cols:

    df[col] = df[col].fillna(
        df[col].median()
    )

print("Numeric missing values filled.")
```

```python
categorical_cols = [
    "Name",
    "Preferred Foot",
    "Position"
]

# Fill categorical missing values using mode
for col in categorical_cols:

    if df[col].isnull().sum() > 0:

        df[col] = df[col].fillna(
            df[col].mode()[0]
        )

print("Categorical missing values filled.")


print("\nRemaining missing values:")

print(df.isnull().sum())

print(
    "\nTotal missing values:",
    df.isnull().sum().sum()
)
```

```python
# IQR Outlier Detection

for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print("\n", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

```python
print("Final dataset shape:", df.shape)

display(df.head(10))
```

```python
# Save cleaned file
df.to_csv(
    "sports_data_cleaned.csv",
    index=False
)

print("Cleaned sports dataset saved successfully.")
```
